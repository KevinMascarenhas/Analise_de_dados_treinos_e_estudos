# 1. Introdução / Análise do Problema

Uma clínica médica tem lidado com uma alta quantidade de sinistros no último ano. A clínica busca identificar quanto gasto com esses sinistros, identificar e filtrar quais pacientes mais contribuem com esses gastos e filtrar e classificar pacientes que podem ter possíveis sinistros futuramente.

```python
import pandas as pd


df_planosaude = pd.read_excel(caminho_dados)

df_planosaude = df_planosaude.loc[:, ~df_planosaude.columns.str.contains('^Unnamed')]

display(df_planosaude)

```

```python
top10_maioresgastos_sinistralidade = df_planosaude.nlargest(15, 'Sinistro_Anual')
display(top10_maioresgastos_sinistralidade)
```

```python
df_planosaude['Sinistralidade'] = (df_planosaude['Sinistro_Anual'] / (df_planosaude['Mensalidade'] * 12)) * 100
df_planosaude['Sinistralidade'] = df_planosaude['Sinistralidade'].round(0).astype(str) + '%'
display(df_planosaude)
```

```python
df_planosaude.sort_values(by='Sinistralidade', ascending=False)
display(df_planosaude)
```

```python
if df_planosaude['Sinistralidade'].dtype == 'object':
    df_planosaude['Sinistralidade'] = df_planosaude['Sinistralidade'].str.replace('%', '').astype(float)

bins = [0, 20, 50, 100, 300]
labels = ['Muito Baixo', 'Médio', 'Alto', 'Crítico']

df_planosaude['Classificacao_Sinistralidade'] = pd.cut(df_planosaude['Sinistralidade'], bins=bins, labels=labels)

print(df_planosaude[['Nome', 'Sinistralidade', 'Classificacao_Sinistralidade']])
```

# 2. Análise Geral

Aqui, tentaremos analisar mais a fundo, analisando cenários de pacientes de risco, pacientes com alta sinistralidade, maior quantidade de internações, entre outros fatores.

```python
alvos = ['Alto', 'Crítico']

df_filtro_risco = df_planosaude[df_planosaude['Classificacao_Sinistralidade'].isin(alvos)]
print(df_filtro_risco)
```

```python
df_planosaude['Sinistro_Anual'] = pd.to_numeric(df_planosaude['Sinistro_Anual'])

Total_Gastos_da_Empresa = df_planosaude['Sinistro_Anual'].sum().astype(int)

Total_Gastos_filtrorisco = df_filtro_risco['Sinistro_Anual'].sum().astype(int)

print(f"Total de Gastos da Empresa: ", Total_Gastos_da_Empresa)

print(f"Total de Gastos de Pacientes com alta sinistralidade:", Total_Gastos_filtrorisco)

# Pacientes em estado "Alto" ou "Crítico" de Sinistralidade são responsáveis por R$ 104.660 gastos com sinistros. Somando com pacientes em estado "Médio" ou "Muito baixo", são R$ 113.330.
```

```python
contagem = df_planosaude['Fumante'].value_counts()
print(contagem)
```

```python
analise_fumo = df_filtro_risco.groupby(['Fumante'])['Sinistro_Anual'].sum()

print(analise_fumo)

# Mesmo havendo somente 4 fumantes entre 15 pacientes, os custos de sinistro com esse grupo chegam a quase 50%.
```

```python
print(f"Encontrados {len(df_filtro_risco)} casos para atenção imediata:")
print(df_filtro_risco[['Nome', 'Idade', 'Sinistralidade', 'Classificacao_Sinistralidade']])
```

```python
media_sinistro_risco = df_filtro_risco['Sinistro_Anual'].mean().astype(int)
media_sinistro_geral = df_planosaude['Sinistro_Anual'].mean().astype(int)

print(f"Média de Sinistro para Pacientes com Alta Sinistralidade: {media_sinistro_risco}")
print(f"Média de Sinistro Geral: {media_sinistro_geral}")
```
